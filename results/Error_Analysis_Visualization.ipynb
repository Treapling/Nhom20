## 🎯 FIGURE 1: Model Performance Comparison (4 Metrics)

```python
fig, axes = plt.subplots(2, 2, figsize=(16, 12))
fig.suptitle('📊 Model Performance Comparison (Đơn vị: Triệu VNĐ)', 
             fontsize=18, fontweight='bold', y=0.995)

# 1.1: MAE Comparison
ax1 = axes[0, 0]
colors_mae = ['red' if x > 30000 else 'orange' if x > 25000 else 'green' 
              for x in df_models['MAE']]
bars1 = ax1.barh(df_models['Model'], df_models['MAE'], color=colors_mae, alpha=0.8, edgecolor='black')
ax1.set_xlabel('MAE (Triệu VNĐ)', fontsize=12, fontweight='bold')
ax1.set_title('1. Mean Absolute Error (MAE)', fontsize=13, fontweight='bold')
ax1.axvline(x=25000, color='green', linestyle='--', linewidth=2, label='Good Threshold')
ax1.legend()
for i, v in enumerate(df_models['MAE']):
    ax1.text(v + 500, i, f'{v:,.0f}', va='center', fontweight='bold', fontsize=9)

# 1.2: R² Score Comparison
ax2 = axes[0, 1]
colors_r2 = ['red' if x < 0 else 'orange' if x < 0.1 else 'green' 
             for x in df_models['R2_Score']]
bars2 = ax2.barh(df_models['Model'], df_models['R2_Score'], color=colors_r2, alpha=0.8, edgecolor='black')
ax2.set_xlabel('R² Score', fontsize=12, fontweight='bold')
ax2.set_title('2. R² Score (Cao càng tốt)', fontsize=13, fontweight='bold')
ax2.axvline(x=0, color='black', linestyle='-', linewidth=1)
ax2.axvline(x=0.1, color='orange', linestyle='--', linewidth=2, label='Target: 0.1+')
ax2.legend()
for i, v in enumerate(df_models['R2_Score']):
    ax2.text(v + 0.005, i, f'{v:.4f}', va='center', fontweight='bold', fontsize=9)

# 1.3: MAPE Comparison
ax3 = axes[1, 0]
colors_mape = ['red' if x > 160 else 'orange' if x > 130 else 'green' 
               for x in df_models['MAPE']]
bars3 = ax3.barh(df_models['Model'], df_models['MAPE'], color=colors_mape, alpha=0.8, edgecolor='black')
ax3.set_xlabel('MAPE (%)', fontsize=12, fontweight='bold')
ax3.set_title('3. Mean Absolute Percentage Error (MAPE)', fontsize=13, fontweight='bold')
ax3.axvline(x=130, color='green', linestyle='--', linewidth=2, label='Good Threshold')
ax3.legend()
for i, v in enumerate(df_models['MAPE']):
    ax3.text(v + 3, i, f'{v:.2f}%', va='center', fontweight='bold', fontsize=9)

# 1.4: Median Error Comparison
ax4 = axes[1, 1]
colors_median = ['red' if x > 5000 else 'orange' if x > 3000 else 'green' 
                 for x in df_models['Median_Error']]
bars4 = ax4.barh(df_models['Model'], df_models['Median_Error'], color=colors_median, alpha=0.8, edgecolor='black')
ax4.set_xlabel('Median Error (Triệu VNĐ)', fontsize=12, fontweight='bold')
ax4.set_title('4. Median Error', fontsize=13, fontweight='bold')
ax4.axvline(x=3000, color='green', linestyle='--', linewidth=2, label='Good Threshold')
ax4.legend()
for i, v in enumerate(df_models['Median_Error']):
    ax4.text(v + 150, i, f'{v:,.0f}', va='center', fontweight='bold', fontsize=9)

plt.tight_layout()
plt.savefig('01_model_performance_comparison.png', dpi=300, bbox_inches='tight')
print("✅ Saved: 01_model_performance_comparison.png")
plt.show()
```

## 🌐 FIGURE 4: Radar Chart - Multi-dimensional Comparison

```python
# Normalize metrics (0-100 scale)
def normalize_inverse(value, min_val, max_val):
    if max_val == min_val:
        return 50
    return ((max_val - value) / (max_val - min_val)) * 100

fig, ax = plt.subplots(figsize=(10, 10), subplot_kw=dict(projection='polar'))

metrics_names = ['MAE (Inverse)', 'RMSE (Inverse)', 'R² Score', 'MAPE (Inverse)']
best_models = ['Stacking', 'Random Forest', 'XGBoost', 'LightGBM']

# Select top models
df_top = df_models[df_models['Model'].isin(best_models)].copy()

angles = [n / len(metrics_names) * 2 * pi for n in range(len(metrics_names))]
angles += angles[:1]

colors_radar = ['#FF6B6B', '#4ECDC4', '#45B7D1', '#FFA07A']

for idx, (model, color) in enumerate(zip(best_models, colors_radar)):
    row = df_top[df_top['Model'] == model].iloc[0]
    
    # Normalize metrics
    mae_norm = normalize_inverse(row['MAE'], df_models['MAE'].min(), df_models['MAE'].max())
    rmse_norm = normalize_inverse(row['RMSE'], df_models['RMSE'].min(), df_models['RMSE'].max())
    r2_norm = (row['R2_Score'] + 0.02) * 500  # Boost R2
    mape_norm = normalize_inverse(row['MAPE'], df_models['MAPE'].min(), df_models['MAPE'].max())
    
    values = [mae_norm, rmse_norm, r2_norm, mape_norm]
    values += values[:1]
    
    ax.plot(angles, values, 'o-', linewidth=2.5, label=model, color=color)
    ax.fill(angles, values, alpha=0.15, color=color)

ax.set_xticks(angles[:-1])
ax.set_xticklabels(metrics_names, size=12, fontweight='bold')
ax.set_ylim(0, 100)
ax.set_yticks([20, 40, 60, 80, 100])
ax.set_rlabel_position(0)
ax.grid(True, linewidth=0.8)
ax.set_title('🌐 Multi-Dimensional Model Performance Radar', 
             fontsize=14, fontweight='bold', pad=20)
ax.legend(loc='upper right', bbox_to_anchor=(1.3, 1.1), fontsize=11)

plt.tight_layout()
plt.savefig('04_radar_chart.png', dpi=300, bbox_inches='tight')
print("✅ Saved: 04_radar_chart.png")
plt.show()
```

## 🔥 FIGURE 5: Heatmap - Performance Matrix

```python
# Normalize all metrics (0-100)
df_norm = df_models.copy()
df_norm['MAE'] = 100 - ((df_norm['MAE'] - df_norm['MAE'].min()) / 
                         (df_norm['MAE'].max() - df_norm['MAE'].min())) * 100
df_norm['RMSE'] = 100 - ((df_norm['RMSE'] - df_norm['RMSE'].min()) / 
                          (df_norm['RMSE'].max() - df_norm['RMSE'].min())) * 100
df_norm['R2_Score'] = (df_norm['R2_Score'] + 0.02) * 500
df_norm['MAPE'] = 100 - ((df_norm['MAPE'] - df_norm['MAPE'].min()) / 
                          (df_norm['MAPE'].max() - df_norm['MAPE'].min())) * 100
df_norm['Median_Error'] = 100 - ((df_norm['Median_Error'] - df_norm['Median_Error'].min()) / 
                                  (df_norm['Median_Error'].max() - df_norm['Median_Error'].min())) * 100

heatmap_data = df_norm[['Model', 'MAE', 'RMSE', 'R2_Score', 'MAPE', 'Median_Error']].set_index('Model')

fig, ax = plt.subplots(figsize=(12, 8))
sns.heatmap(heatmap_data, annot=True, fmt='.1f', cmap='RdYlGn', center=50,
            cbar_kws={'label': 'Performance Score (0-100)'}, 
            linewidths=0.5, linecolor='gray', ax=ax)

ax.set_title('🔥 Performance Heatmap (Normalized Score)\n(Green=Better, Red=Worse)', 
             fontsize=14, fontweight='bold', pad=15)
ax.set_xlabel('Metrics', fontsize=12, fontweight='bold')
ax.set_ylabel('Models', fontsize=12, fontweight='bold')

plt.tight_layout()
plt.savefig('05_heatmap_matrix.png', dpi=300, bbox_inches='tight')
print("✅ Saved: 05_heatmap_matrix.png")
plt.show()
```
